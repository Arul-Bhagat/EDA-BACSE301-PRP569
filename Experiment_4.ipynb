```python
import numpy as np
import pandas as pd
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer, SimpleImputer
df = pd.DataFrame(
 {
 "Roll_no": [24, np.nan, 31, 28, 45],
 "Marks": [92, 83, np.nan, 83, 86],
 "Grade": ["A", "B", "A", np.nan, "B"],
 }
)
mean_imputer = SimpleImputer(strategy="mean")
df["Roll_no_Mean"] = mean_imputer.fit_transform(df[["Roll_no"]])
print(df)
```

```python
# 2. Median Imputation (Numerical - robust to outliers)
median_imputer = SimpleImputer(strategy="median")
df["Marks_Median"] = median_imputer.fit_transform(df[["Marks"]])
print(df)
```

```python
# 3. Mode Imputation (Categorical - most frequent)
mode_imputer = SimpleImputer(strategy="most_frequent")
df["Grade_Mode"] = mode_imputer.fit_transform(df[["Grade"]]).ravel()
print(df)
```

```python
# 4. MICE Imputation (Multivariate Imputation by Chained Equations)
# Select numerical columns for MICE
num_cols = ["Roll_no", "Marks"]
mice_imputer = IterativeImputer(max_iter=10, random_state=42)
df_mice = df[num_cols].copy()
df_mice.iloc[:, :] = mice_imputer.fit_transform(df_mice)
df["Roll_no_MICE"] = df_mice["Roll_no"]
df["Marks_MICE"] = df_mice["Marks"]
print(df)
```
