```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df= pd.read_csv("airline_passengers.csv")
zero_mean_series = np.random.normal(loc=0.0, scale=1., size=50)
zero_mean_series
```

```python
plt.figure(figsize=(16, 8))
g = sns.lineplot(data=zero_mean_series)
zero_mean_series
```

```python
sns.set(rc={'figure.figsize':(11, 4)}) #Runtime Configuration
plt.rcParams['figure.figsize'] = (8,5)
plt.rcParams['figure.dpi'] = 150
df['total_passengers'].plot(linewidth=0.25)
```

```python
cols_to_plot = ['total_passengers']
axes = df[cols_to_plot].plot(marker='.', alpha=0.5,linestyle='None',figsize=(14, 6), subplots=True)
axes
```

```python
ax = df.loc['1952', 'total_passengers'].plot()
ax.set_ylabel('Total Passengers');
```

```python
columns = ['total_passengers']
passengers_yearly_mean = df[columns].resample('YE').mean()
passengers_yearly_mean
```

```python
start, end = '1950','1955'
fig, ax = plt.subplots()
ax.plot(df.loc[start:end, 'total_passengers'], marker='.', linestyle='-', linewidth=0.5, label='Monthly')
ax.plot(passengers_yearly_mean.loc[start:end, 'total_passengers'], marker='*', markersize=8, linestyle='-', label='Yearly Mean Resample')
ax.set_ylabel('Total Passengers')
ax.legend();
```
