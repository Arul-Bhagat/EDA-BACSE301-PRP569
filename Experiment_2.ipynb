```python
import pandas as pd
df_C_marketing= pd.DataFrame({'emp_ID': [2345,6743,6854,3462],'Emp_name': ['A','B','C','D']})
df_C_sales= pd.DataFrame({'emp_ID':[2654,7892,3679,5432],'Emp_name':['E','F','G','H']})
df_R_chemistry= pd.DataFrame({'emp_ID': [3785,9834,1002,4985],'Emp_name':['I','J','K','L']})
df_R_physics= pd.DataFrame({'emp_ID': [4800,923,4572,8003],'Emp_name':['M','N','O','P']})
dfC = pd.concat([df_C_marketing,df_C_sales], ignore_index=True)
dfR = pd.concat([df_R_chemistry,df_R_physics], ignore_index=True)

df = pd.concat([dfC,dfR], axis=1)
df
```
